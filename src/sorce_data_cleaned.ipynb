## Training points pikachu cleaned

```python
data_path_training_pikachu = "../data_training"
path_training_pikachu = f"{data_path_training_pikachu}/pikachu.txt"


import numpy as np

data_points = []

with open(path_training_pikachu, "r") as f_read, open(f"{data_path_training_pikachu}/pikachu_cleaned.txt", "w") as f_write:  
 
 for i, pikachu in enumerate(f_read):
  if i > 0:
        pikachu = pikachu.strip()

        if pikachu != "":
            pikachu = pikachu.strip("()")
            width_pik, height_pik = pikachu.split(",")
            width_pik = float(width_pik)
            height_pik = float(height_pik)
            data_point = [width_pik, height_pik]
            data_points.append(data_point)
    
    
        f_write.write(f"{pikachu}\n")
        data_points_array = np.array(data_points)

width_pik = data_points_array[:,0]
height_pik = data_points_array[:,1]

```

## Training data pichu cleaned

```python


data_path_training_pichu = "../data_training"


path_training_pichu = f"{data_path_training_pichu}/pichu.txt"

import numpy as np
pichu_coords=[]

with open(path_training_pichu, "r") as f_read, open (f"{data_path_training_pichu}/pichu_cleaned.txt", "w") as f_write:
    
    
    for row in f_read:
        
        row = row.strip()
        if  "(width" in row:
            continue
        
                
        row = row.strip("()")
        separate_coords = row.split(",") #separate by the comma
        width_pic=float(separate_coords[0])
        width_pic = round(width_pic,2) #rounding of width to two decimals
        height_pic=float(separate_coords[1]) 
        height_pic = round(height_pic,2) #rounding of height to two decimals
        pichu_coords.append([width_pic, height_pic]) 
    
    pichu_coords = np.array(pichu_coords)
    f_write.write(f"{pichu_coords}")

width_pic = pichu_coords[:,0]
height_pic = pichu_coords[:,1]


# plot
      
import matplotlib.pyplot as plt
       

plt.figure(figsize=(10,10))
plt.scatter(pichu_coords[:,0], pichu_coords[:,1], color='blue' ) 
plt.scatter(data_points_array[:,0], data_points_array[:,1], color='red' )   

#print(pichu_coords)

    
```

## user data cleaned

```python
import numpy as np
import re
import ast

data_path_test_points = "../data_test_points"
path_test_points = f"{data_path_test_points}/test_points.txt"

test_points_coords = []


with (open(path_test_points, "r") as f_read, open(f"{data_path_test_points}/test_points_cleaned.txt", "w") as f_write,):

    for row in f_read:
        row = row.strip()
        row = re.sub(" +", "", row)
        row = row.replace("(","").replace(")","")
             
        
        row = ast.literal_eval(row)
        #row = list(row)
        for i in range(0, len(row), 2):
            column1 = float(row[i])
            column2 = float(row[i + 1])

            test_points_coords.append([column1, column2])   
         

        test_points_coords = np.array(test_points_coords)
    
    f_write.write(f"{test_points_coords}")
    

width_t=test_points_coords[:,0]
height_t=test_points_coords[:,1]
#print(width_t)
#print(height_t)
    
```
